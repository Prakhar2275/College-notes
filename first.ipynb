Traditional Programming code.......


```python
from typing import List
def calc(selling_p: List[float],costp:List[float])->List[float]:
  assert len(selling_p)==len(costp)
  profits=[]

  for i in range(len(costp)):
    profits.append(selling_p[i]-costp[i])

  return profits

ans=calc(list([70,80,100]),list([60,70,90]))

for i in ans:
  print(i)
```
